# Attack evaluations

Attack scenario: the attacker can manipulate the input of each member of the ensemble. Each model is preferentially given adversarial samples that were generated against it on the test dataset, to maximize effectiveness. Tree-based models do not support white-box attacks, so the source of white-box adversarial inputs for them is the Deep Learning models.
+ For tree-based models, the attacks are ZOO and HSJA (black-box attacks); gradient-based white-box attacks are not applicable to them directly.
+ For deep learning models, besides ZOO and HSJA, the white-box attacks (FGSM, PGD, JSMA, C&W, DeepFool) are run. Since there are two kinds of DL models, both transfer scenarios are evaluated: from DNN and from TabNet.
+ Variants with adversarial training (robust) are repeated over 4 pool seeds (1337, 1338, 1339, 1340) via `--model-seeds`; each result cell is mean ± std over both the model seed and the attack data source (`--model-seeds` combined with `--seed-dirs`). The isolated variant has a single checkpoint per model and is not repeated over model seeds. Variants without adversarial training keep the original model.

# Transfer demonstrations: explicit source → target

The evaluations above choose the adversarial input of each model automatically: a model uses the vectors generated against itself when they exist (DNN, TabNet, and the ZOO/HSJA pools of the trees), and `--adv-fallback-src {dnn,tabnet}` only fills in the gradient-based attacks that a tree cannot generate. Consequently DNN and TabNet never receive each other's vectors there, and no tree-generated vector reaches a DL model.

The cells below measure single pathways `(source, attack, target)` explicitly. `--adv-in` passes the source's adversarial CSV (the source is encoded in the file name) and `-m` selects the target, so the evaluator applies the unchanged vectors without any re-optimization on the target. In every table, the column is the target and the row is the file, i.e. the source and attack; the `original` row is the clean accuracy of the target.

+ **DL ↔ DL**: TabNet → DNN and DNN → TabNet for the five gradient-based attacks (within-family transfer between the two networks, both directions).
+ **Tree → DL**: the ZOO/HSJA vectors generated against each tree (XGB, CatBoost, Bagging, GBM) evaluated on DNN and TabNet. Here the source is a black-box query attack, the target a network.
+ **Reference**: the direct case (source = target) is the `Individual models: No defenses` section above.

A cell value `acc (asr: x)` is the accuracy on the transferred vectors and the attack success rate on the records the target classified correctly when clean. All cells use the no-defense checkpoints.
